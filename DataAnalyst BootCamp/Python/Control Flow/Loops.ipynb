## Loops
A loop is used to execute a block of code repeatedly until a particular condition is met.

Imagine you want to print "Hello" 10 times. Instead of writing print("Hello") ten times, you can use a loop.

Python has two main types of loops:
1. for loop
2. while loop

```python
## for loop
for i in range(5):
    print(i)
```

```python
for i in range(1,10,2):
    print(i)
```

```python
for i in range(10,1,-1):
    print(i)
```

```python
for i in range(10,1,-2):
    print(i)
```

```python
## Strings
str="Saquib Bin Halim"
for i in str:
    print(i)
```
