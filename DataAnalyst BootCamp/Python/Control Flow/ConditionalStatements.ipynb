## Conditional Statements
Conditional statements are used to make decisions in a program. They execute different blocks of code depending on whether a given condition is True or False.

In simple words, conditional statements help Python decide what to do based on a condition.

Important concepts to remember
- if checks a condition.
- elif means else if.
- else executes when all previous conditions are false.
- : indicates the beginning of a code block.
- Indentation defines which statements belong to a block.
- Conditions commonly use comparison and logical operators.

Statement   -  Purpose
* if	-   Checks one condition
* if-else	-   Chooses between two outcomes
* if-elif-else  -     Checks multiple conditions
* Nested if - 	Checks a condition inside another condition

```python
## if statement

age=25
if age>=20:
    print("adult, start earning your own money")
```

```python
## if - else statement
## the else statement executes in the block of code if the "if" statement is false
age=16
if(age>=18):
    print("You are eligible to vote")
else:
    print("You are not eligible grow up!!")
    
```

```python
## elif
## it allows you to check multiple conditions , It stands for "else-if"

age=int(input("Enter your age: "))
if age<13:
    print("You are a child")
elif age<18:
    print("You are a teen ager")
else:
    print("You are an adult")
```

## Nested Conditional Statements
A nested conditional statement means writing an if statement inside another if or else statement.

In simple words, Python checks one condition first. If that condition is satisfied, it checks another condition inside it.

```python
## even , odd , negative , positive

num=int(input("Enter the Number: "))

if num>0:
    print("Number is Positive")
    if num%2==0:
        print("Number is Even")
    else:
        print("Number is Odd")
else:
    print("Number is Negative")
```

```python
## Leaf Year Problem

# Determine if a year is leap year or not , using Nested Conditional statements

year=int(input("Enter the Year: "))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year," is a leap year")
        else:
            print(year," is not a leap year")
    else:
        print(year," is a leap year")
else:
    print(year," is not a leap year")
```

A normal year has 365 days. A leap year has 366 days.

But not every year divisible by 4 is automatically a leap year.

There are three rules:
1. If a year is divisible by 400, it is a leap year.
2. If a year is divisible by 100, but not by 400, it is NOT a leap year.
3. Otherwise, if it is divisible by 4, it is a leap year.
