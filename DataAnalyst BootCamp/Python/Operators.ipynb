Python has seven main operator categories: Arithmetic, Comparison, Assignment, Logical, Bitwise, Membership and Identity.

1. Arithmetic Operators
Definition: Used to perform mathematical calculations such as addition, subtraction, multiplication and division.

```python
fruits={"apple","Banana","PineApple"}
fruits.add("Orange")
print(fruits)
```

```python
a=10
b=15

add_result=a+b
sub_result=a-b
div_result=a/b
mul_result=a*b
floor_result=a//b
exponent_result=a**b
modulus_result=a%b

print(add_result)
print(sub_result)
print(mul_result)
print(floor_result)
print(exponent_result)
print(modulus_result)
print(div_result)
```
