Understanding Variable Types:
1. Python is Dynamically Typed:
Type of a variable is determined at Run-time



```python
## Type Conversion
age=25
print(age)
print(type(age))

age_str=str(age)
print(age_str)
print(type(age_str))
```

```python
height=5.6
type(height)
```

```python
float(int(height))
```

Dynamic Typing: Dynamic typing in Python means you don't need to declare a variable's data type explicitly. Python automatically determines the type based on the value assigned to it.


```python
var=10
print(var,type(var))
# python allows the type of a variable to change as the program executes

var="Saquib"
print(var,type(var))

var=3.14
print(var,type(var))
```
