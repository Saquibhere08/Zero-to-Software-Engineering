## Numpy: 
NumPy (Numerical Python) is a Python library used for performing mathematical and numerical operations efficiently, especially when working with large amounts of data.

It is one of the most important libraries for Data Analytics.
## Why do we use NumPy?
NumPy helps us to:
- Perform mathematical calculations.
- Work with arrays and matrices.
- Perform statistical operations.
- Handle large numerical datasets.
- Perform calculations faster than ordinary Python lists in many numerical workloads.

```python
import numpy as np

#1-D array created with numpy
arr1=np.array([1,2,3,4,5])
print(arr1)
print(type(arr1))
print(arr1.shape) # it will provide single dimension array
```
